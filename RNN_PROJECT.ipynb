# data preprocessing

```python
quotes = quotes.str.lower()
```

```python
import string
translator = str.maketrans('', '', string.punctuation)
quotes = quotes.apply(lambda x: x.translate(translator))
quotes.head()
```

# Tokenization

```python
vocab_size = 8979
from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer = Tokenizer(num_words=vocab_size)
tokenizer.fit_on_texts(quotes)
```

```python
word_index = tokenizer.word_index
print(len(word_index))
list(word_index.items())[:10]
```

```python
sequence = tokenizer.texts_to_sequences(quotes)
```

```python
quotes[0]
```

```python
sequence[0]
```

# input and output


```python
X = []
y = []

for seq in sequence:
    for i in range(1, len(seq)):
      input_seq = seq[:i]
      output_seq = seq[i]
      X.append(input_seq)
      y.append(output_seq)
```

```python
# X
```

```python
len(X)
```

```python
len(y)
```

# padding

```python
max_len = max(len(x) for x in X)
print(max_len )
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
X_padded = pad_sequences(X , maxlen = max_len, padding = 'pre')
```

```python
X_padded[0]
```

```python
y = np.array(y)
```

```python
X_padded.shape
```

# one hot encoding on y

```python
y.shape
```

```python
y[0]
```

```python
from tensorflow.keras.utils import to_categorical
y_one_hot = to_categorical(y,num_classes = vocab_size)
```

```python
y_one_hot[0]
```

```python
y_one_hot.shape
```

# Simple RNN model

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN,LSTM, Dense
```

```python
embedding_dim = 20
rnn_units = 128
```

```python
rnn_model = Sequential()

rnn_model.add(Embedding(
        input_dim = vocab_size,
        output_dim = embedding_dim,
        input_length = max_len))

rnn_model.add(SimpleRNN(
    units = rnn_units,
    return_sequences = False))

rnn_model.add(Dense(
    units=vocab_size,
    activation = 'softmax'
    ))
```

```python
rnn_model.compile(
    optimizer = 'adam',
    loss = 'categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
rnn_model.summary()
```

# LSTM model

```python
lstm_model = Sequential()

lstm_model.add(Embedding(
        input_dim = vocab_size,
        output_dim = embedding_dim,
        input_length = max_len))

lstm_model.add(LSTM(
    units = rnn_units,
    return_sequences = False))

lstm_model.add(Dense(
    units = vocab_size,
    activation = 'softmax'))
```

```python
lstm_model.compile(
    optimizer = 'adam',
    loss = 'categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
lstm_model.summary()
```

# RNN model training

```python
eposch = 10
batch_size = 128

history_rnn = rnn_model.fit(
    X_padded, y_one_hot,
    epochs = eposch,
    batch_size = batch_size,
    validation_split = 0.1

)
```

 geting less accuracy due to restricted performance ,  accuracy will increase if we increase the number of epochs.

## using saved model for predicting next word

```python
from tensorflow.keras.models import load_model

model = load_model('lstm_model.h5')
model.summary()
```

```python
downloaded_lstm_model = model
```

```python
index_to_word = dict((i, word) for word, i in tokenizer.word_index.items())
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
def predictor(model ,tokenizer,text , max_len):

  text = text.lower()

  seq = tokenizer.texts_to_sequences([text])
  seq = pad_sequences(seq , maxlen = max_len, padding = 'pre')

  pred = model.predict(seq , verbose = 0)
  pred_index = np.argmax(pred)

  pred_word = index_to_word[pred_index]
  return pred_word
```

```python
seed_text = "life is"
next_word = predictor(downloaded_lstm_model, tokenizer, seed_text, max_len)
print(next_word)
```

```python
seed_text = "what are you"
next_word = predictor(downloaded_lstm_model, tokenizer, seed_text, max_len)
print(next_word)
```

# now using saved model for predicting the 'n' next words

```python
def generate_text(model,tokenizer,seed_text,max_len,n_words):

  for _ in range(n_words):
    next_word = predictor(model,tokenizer,seed_text,max_len)
    if next_word == "":
      break
    seed_text += " " + next_word

  return seed_text
```

```python
seed = "are you a "
next_words = generate_text(downloaded_lstm_model,tokenizer,seed,max_len,10)
print(next_words)
```

```python
seed = "It is our choices, Harry, that  "
next_words = generate_text(downloaded_lstm_model,tokenizer,seed,max_len,10)
print(next_words)
```
